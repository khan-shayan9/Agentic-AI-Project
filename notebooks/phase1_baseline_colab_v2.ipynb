## 8. Choose the verifier prompt on the validation split

Three prompt styles are compared on the 200 validation answers (never on test):
- **digits**: reply 0, 1 or 2
- **words**: reply FACTUAL, HALLUCINATED or UNSURE
- **reason**: one or two sentences of checking, then a last line `Verdict: ...`

PubMed results are cached, so only the first style searches PubMed. The table shows the vote and each verifier, with how often each answered 0 / 1 / 2. Pick the style with the best vote F1 and accuracy on answered items, and no verifier stuck on one answer.

The first run takes about 80 minutes on a T4; finished items are saved, so running it again is quick. Result: `reason` won (vote F1 0.683, coverage 0.90) and is now the default.

```python
for style in ["digits", "words", "reason"]:
    !python -m medverify pipeline --split val --evidence pubmed --prompt {style} --name sysA-val-{style} > /dev/null
!python -m medverify compare results/runs/sysA-val-digits results/runs/sysA-val-words results/runs/sysA-val-reason
```

## 9. Choose the vote rule on the validation split

`tune` tries every rule from 1 to 3 agreeing verifiers and keeps the one with the best accuracy on answered items that still answers at least 70% of them. `overlap` shows whether the verifiers make the same mistakes. Use the chosen rule as `--min-agree` for the test run.

```python
!python -m medverify tune results/runs/sysA-val-reason --min-coverage 0.7
!python -m medverify overlap results/runs/sysA-val-reason
```

## 10. Numbers for the proposal's results table

```python
import json, pathlib
for run in ["llama-test-none", "llama-test-source", "sysA-dev"]:
    path = pathlib.Path("results/runs") / run / "metrics.json"
    if not path.exists():
        print(f"{run}: not run yet")
        continue
    m = json.loads(path.read_text())
    o, ci = m["overall"], m["overall"].get("f1_ci95")
    ci_text = f" ({ci[0]:.3f}-{ci[1]:.3f})" if ci else ""
    levels = "  ".join(f"{lvl} F1 {m[lvl]['f1']:.3f}" for lvl in ("easy", "medium", "hard") if lvl in m)
    print(f"{run}: precision {o['precision']:.3f}  recall {o['recall']:.3f}  F1 {o['f1']:.3f}{ci_text}  coverage {o['coverage']:.3f}  {levels}")
```
